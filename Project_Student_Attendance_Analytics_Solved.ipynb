## Student Attendance Analytics with Pandas

```python
import pandas as pd
import numpy as np
```

#### Section 0: Data Loading

```python
df = pd.read_excel("aiml_attendance.xlsx")
df
```

#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""

print("First 10 rows:")
display(df.head(10))

print("DataFrame Shape:")
print(df.shape)

print("Data Types:")
print(df.dtypes)
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count."""

# Treat the text "Never" as a missing value for the date column
df["Last Present Date"] = df["Last Present Date"].replace("Never", np.nan)

print("Missing values in each column:")
print(df.isnull().sum())

print("\nColumns containing missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""

df.rename(columns={
    "Roll No": "roll_no",
    "Student Name": "student_name",
    "Subject": "subject",
    "Total Classes": "total_classes",
    "Present": "present",
    "Absent": "absent",
    "Excused": "excused",
    "Attendance %": "attendance_pct",
    "Status": "status",
    "Classes Needed (75%)": "classes_needed_75",
    "Last Present Date": "last_present_date",
    "Consecutive Absences": "consecutive_absences"
}, inplace=True)

print("Updated column names:")
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""

df["attendance_pct"] = (
    df["attendance_pct"].astype(str).str.replace("%", "", regex=False).str.strip()
)
df["attendance_pct"] = pd.to_numeric(df["attendance_pct"], errors="coerce") / 100

df["last_present_date"] = pd.to_datetime(df["last_present_date"], errors="coerce")

print(df[["attendance_pct", "last_present_date"]].head())
print("\nUpdated data types:")
print(df[["attendance_pct", "last_present_date"]].dtypes)
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""

critical_students = df[df["status"].str.strip().str.lower() == "critical"]
critical_students = critical_students.sort_values("attendance_pct", ascending=True)
display(critical_students)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""

fully_absent = df[df["present"] == 0][
    ["roll_no", "student_name", "consecutive_absences"]
]
display(fully_absent)
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """

print("(a) Safe students using .loc:")
display(df.loc[df["status"].str.strip().str.lower() == "safe"])

print("(b) First 15 rows and columns 2-6 using .iloc:")
display(df.iloc[:15, 1:6])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""

students_absent_5_or_more = df.loc[
    df["consecutive_absences"] >= 5,
    ["student_name", "status", "last_present_date"]
]
display(students_absent_5_or_more)
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute:
a) Count of students
b) Average attendance %
c) Maximum consecutive absences"""

status_summary = df.groupby("status").agg(
    student_count=("roll_no", "count"),
    average_attendance_pct=("attendance_pct", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)
display(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""

df["attendance_category"] = np.select(
    [
        df["attendance_pct"] >= 0.90,
        df["attendance_pct"] >= 0.75,
        df["attendance_pct"] >= 0.60
    ],
    ["Excellent", "Good", "Average"],
    default="Poor"
)

print("Student count in each attendance category:")
display(df.groupby("attendance_category").size().to_frame("student_count"))
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""

totals = df.agg({
    "total_classes": "sum",
    "present": "sum",
    "absent": "sum"
})
print(totals)
```

```python
"""Q12. Group by Status and use .agg() to find:
a) min, max, and mean of Present
b) sum of Absent"""

status_attendance = df.groupby("status").agg(
    present_min=("present", "min"),
    present_max=("present", "max"),
    present_mean=("present", "mean"),
    absent_sum=("absent", "sum")
)
display(status_attendance)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""

df["risk_score"] = (df["absent"] * 0.4) + (df["consecutive_absences"] * 0.6)
top_10_risk = df.sort_values("risk_score", ascending=False).head(10)
display(top_10_risk[["roll_no", "student_name", "absent", "consecutive_absences", "risk_score"]])
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""

df["classes_needed_numeric"] = (
    df["classes_needed_75"].astype(str).str.extract(r"(\d+)")[0].fillna("0").astype(int)
)
display(df[["classes_needed_75", "classes_needed_numeric"]].head(10))
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a) "Defaulter" if attendance < 75%
b) "Safe" otherwise """

df["attendance_flag"] = df["attendance_pct"].apply(
    lambda attendance: "Defaulter" if attendance < 0.75 else "Safe"
)
display(df[["student_name", "attendance_pct", "attendance_flag"]].head(10))
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""

df["attendance_range"] = pd.cut(
    df["attendance_pct"] * 100,
    bins=[0, 50, 75, 90, 100],
    labels=["Critical", "Warning", "Good", "Excellent"],
    include_lowest=True,
    
    right=True
)
display(df[["student_name", "attendance_pct", "attendance_range"]].head(10))
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""

attendance_pivot = pd.pivot_table(
    df,
    index="status",
    columns="attendance_category",
    values="roll_no",
    aggfunc="count",
    fill_value=0
)
display(attendance_pivot)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""


df["base_name"] = (
    df["student_name"].astype(str).str.strip().str.replace(r"\s*\(.*\)$", "", regex=True)
    .str.split().str[0].str.upper()
)
repeated_base_names = df[df.duplicated("base_name", keep=False)].sort_values("base_name")
display(repeated_base_names)
```

```python
"""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the result."""

# Hypothetical details: update Branch/Semester values if your actual details are different.
Student_Details = pd.DataFrame({
    "roll_no": df["roll_no"],
    "branch": ["AI&ML"] * len(df),
    "semester": [5] * len(df)
})

merged_df = pd.merge(df, Student_Details, on="roll_no", how="inner")
display(merged_df.head(10))
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""


critical_students = df[df["status"].str.strip().str.lower() == "critical"]
critical_students.to_csv("critical_students.csv", index=False)



summary_by_status = df.groupby("status").agg(
    student_count=("roll_no", "count"),
    average_attendance_pct=("attendance_pct", "mean"),
    total_present=("present", "sum"),
    total_absent=("absent", "sum"),
    max_consecutive_absences=("consecutive_absences", "max")
)
summary_by_status.to_csv("summary_by_status.csv")


top_10_risk = df.sort_values("risk_score", ascending=False).head(10)
top_10_risk.to_csv("top_10_risk_score.csv", index=False)

print("Exported: critical_students.csv, summary_by_status.csv, top_10_risk_score.csv")
```
